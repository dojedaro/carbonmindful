# CarbonMindful: CPU and GPU comparison (E3)

This runs the same CNN on CIFAR-10 on one machine, once on the CPU and once on the GPU, with five seeds each.

**To run**
1. Choose **Runtime > Change runtime type > T4 GPU**.
2. Choose **Runtime > Run all**, and allow access to Google Drive when asked.
3. When a **Choose Files** button appears below the first code cell, select `code_for_colab.zip`.
4. Keep this tab open. The CPU runs are slow, so this may take a few hours.

**If the session disconnects**, choose **Run all** again. Finished runs are saved in Drive and skipped, and you will not need to select the file again.

When it finishes, the results are in `carbonmindful_e3/runs` in Drive.

```python
from pathlib import Path
import shutil
import torch
from google.colab import drive, files

assert torch.cuda.is_available(), (
    "No GPU: choose Runtime > Change runtime type > T4 GPU, then Run all."
)
print("GPU:", torch.cuda.get_device_name(0))

drive.mount("/content/drive")
FOLDER = Path("/content/drive/MyDrive/carbonmindful_e3")
FOLDER.mkdir(parents=True, exist_ok=True)

CODE = FOLDER / "code_for_colab.zip"
if not CODE.exists():
    print("Select code_for_colab.zip")
    uploaded = files.upload()
    assert "code_for_colab.zip" in uploaded, (
        "Please select the file named code_for_colab.zip, then Run all again."
    )
    shutil.copy("code_for_colab.zip", CODE)
print("Ready.")
```
